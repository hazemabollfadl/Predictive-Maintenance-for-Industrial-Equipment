```python
import importnb
import numpy as np
import pandas as pd
with importnb.Notebook():
    import acquisition_and_profiling as phase_1
    df_train = phase_1.df_train
    engine_lifespans = phase_1.engine_lifespans
```

```python
# Extracting RUL

#Map The engine life span to the each Engine_ID
max_cycles_per_row = df_train.index.get_level_values('Engine_ID').map(engine_lifespans)
# max_cycles_per_row

#Calculate and add RUL
df_train['RUL'] = max_cycles_per_row - df_train.index.get_level_values('Cycle')
df_train
```

```python
# Create and set 'W' threshold for classification
# Use numpy.where() to strictly enforce the binary threshold.
df_train['Label_30'] = np.where(df_train['RUL'] <= 30, 1, 0)

# Print the tail of Engine 1 for visual confirmation.
print(df_train.loc[:, ['RUL', 'Label_30']].tail(5))
```
